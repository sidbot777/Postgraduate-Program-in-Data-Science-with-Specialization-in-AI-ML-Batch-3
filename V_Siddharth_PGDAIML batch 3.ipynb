```python
# 2 Univariate analsys across numerical features
```

```python
num_cols=df.select_dtypes(include=np.number).columns
print(num_cols)
```

```python
#histograms
for col in num_cols:
  plt.figure(figsize=(10,6))
  sns.histplot(df[col],kde=True)
  plt.title(f'Distribution of {col}')
  plt.xlabel(col)
  plt.ylabel('Frequency')
  plt.show()
  print(df[num_cols].skew())
```

```python
#Boxplot
for col in num_cols:
    plt.figure(figsize=(6,4))
    sns.boxplot(x=df[col])
    plt.title(f'Boxplot of {col}')
    plt.show()
```

```python
#interpretation of boxplot
```

```python
#2.i Row.ID  and 记录数 were useless . Year also had limited analytical value beceause oly few years are present.
```

```python
#2.ii #weeknum appears moderately uniform because values are spread across weeks.
#Most business-related variables are not normally distributed.
#Sales and Profit are highly skewed.
```

```python
#2.iii Answer:
#Sales is highly right-skewed.
#Profit is right-skewed.
#Shipping.Cost is right-skewed.
#Quantity is moderately right-skewed.
#Interpretation:
#Right skewness means most orders are small, while a few orders are extremely large.
#This is common in retail datasets.
```

```python
#2.iv Sales contains many extreme outliers.
#Profit contains large positive and negative outliers.
#Shipping.Cost also has several outliers.
#Impact:
#Outliers can affect mean values and machine learning models.
#Median and IQR become more reliable measures.
```

```python
#Bivariate analysis of Numerical Features
corr = df[num_cols].corr()

plt.figure(figsize=(10,6))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

```python
#4.ii Discount and Profit are negatively correlated.
# Higher discounts reduce profit margins.
```
