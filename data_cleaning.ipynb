# Data Cleaning Notebook

Cleans and prepares Hero FinCorp datasets.

```
import pandas as pd
import numpy as np

# Load datasets
customers = pd.read_csv('customers.csv')
loans = pd.read_csv('loans.csv')
transactions = pd.read_csv('transactions.csv')
defaults = pd.read_csv('defaults.csv')
branches = pd.read_csv('branches.csv')

print("Data Loaded Successfully")
```

```
# Remove duplicates
customers.drop_duplicates(inplace=True)
loans.drop_duplicates(inplace=True)
transactions.drop_duplicates(inplace=True)

# Handle missing values
customers.fillna(method='ffill', inplace=True)
loans.fillna(loans.median(numeric_only=True), inplace=True)
transactions.fillna(0, inplace=True)

print("Missing values handled")
```

```
# Convert date columns
transactions['Transaction_Date'] = pd.to_datetime(transactions['Transaction_Date'], errors='coerce')

print("Date formats standardized")
```

```
# Save cleaned datasets
customers.to_csv('clean_customers.csv', index=False)
loans.to_csv('clean_loans.csv', index=False)
transactions.to_csv('clean_transactions.csv', index=False)

print("Cleaned data saved")
```
