```python
#Q1. Filter Altona Data 
altona = df[df['Suburb']== 'Altona']
print(altona['Price'].describe())
```

```python
#Hypotheses

#H0: Mean price = 800000

#H1: Mean price > 800000
```

```python
stat, p_value = ttest_1samp(altona['Price'], popmean= 800000)

print('Test Statistic:', stat)
print('P-value:', p_value)
```

```python
alpha = 0.05

if p_value/2 < alpha and stat > 0:
    print('Reject Null Hypothesis')
    print('Average property price has increased above $800,000')
else:
    print('Fail to Reject Null Hypothesis')
```

```python
#Q2 summer vs winter property prices (2016)
```

```python
sales_2016 = df[df['Year'] == 2016] 
```

```python
winter_months = [10, 11, 12, 1, 2, 3]
sales_2016['Season'] = sales_2016['Month'].apply(lambda x: 'Winter' if x in winter_months else 'Summer')
```

```python
winter_prices = sales_2016[sales_2016['Season'] == 'Winter']['Price']

summer_prices = sales_2016[sales_2016['Season'] == 'Summer']['Price']
```

```python
#Hypotheses
#H0: No difference in average prices

#H1: There is a difference in average prices
```

```python
stat, p_value = ttest_ind(winter_prices, summer_prices, equal_var=False)

print('Test Statistic:', stat)
print('P-value:', p_value)
if p_value < alpha:
    print('Reject Null Hypothesis')
    print('There is a significant difference between summer and winter prices')
else:
    print('Fail to Reject Null Hypothesis')
```

```python
#Q8 Problem Statement

#Check whether Price depends on:

#Suburb
#Type
#Interaction between Suburb and Type
```

```python
# Import libraries

import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols

# Select required columns

anova_df = df[['Suburb', 'Type', 'Price']].copy()

# Remove missing values

anova_df.dropna(inplace=True)

# Convert to categorical datatype

anova_df['Suburb'] = anova_df['Suburb'].astype('category')
anova_df['Type'] = anova_df['Type'].astype('category')

# OPTIONAL:
# Reduce suburbs if dataset is too large
# Keep only top 10 suburbs

top_suburbs = anova_df['Suburb'].value_counts().head(10).index

anova_df = anova_df[anova_df['Suburb'].isin(top_suburbs)]

# Fit Two-Way ANOVA model

model = ols(
    'Price ~ C(Suburb) + C(Type) + C(Suburb):C(Type)',
    data=anova_df
).fit()

# Generate ANOVA table

anova_table = sm.stats.anova_lm(model, typ=2)

# Display result

print(anova_table)
```

```python
#Final Business Insight
#Suburb significantly affects property prices.
#Interaction between Suburb and Type is also significant.
#Property Type alone could not be independently estimated because of data imbalance/multicollinearity.
#Different property types behave differently across different suburbs.
```
